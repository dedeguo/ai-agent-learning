# 02 概率与 softmax。


```python
import numpy as np

scores = np.array([2.0,1.0,0.0])
exp_scores = np.exp(scores)

print("exp_scores \n",exp_scores)
probabilities = exp_scores / exp_scores.sum()

print(probabilities)
print(probabilities.sum())
```

```python
scores_2 = np.array([2, 2, 2])
exp_scores_2 = np.exp(scores_2)

print("scores_2 \n",exp_scores_2)
probabilities_2 = exp_scores_2 / exp_scores_2.sum()

print(probabilities_2)
print(probabilities_2.sum())
```

```python
target =2
loss = - np.log(probabilities[target])
print("loss:",loss)
```

```python
model_a = np.array([0.1,0.8,0.1])
model_b = np.array([0.3,0.4,0.3])
exp_score_a = np.exp(model_a)
exp_score_b = np.exp(model_b)

probabilities_a = exp_score_a / exp_score_a.sum()
probabilities_b = exp_score_b / exp_score_b.sum()


softmax_a=- np.log(probabilities_a[1])
softmax_b=- np.log(probabilities_b[1])

print("softmax_a:",softmax_a)
print("softmax_b:",softmax_b)
```
