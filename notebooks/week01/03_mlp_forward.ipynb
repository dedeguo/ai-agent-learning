# 线性层、激活函数与神经网络的前向计算

我们继续第一周第三课：**线性层、激活函数与神经网络的前向计算**。

前面你已经理解了：

```text
模型分数 → softmax → 概率 → 与真实标签比较 → loss
```

今天来学习：**模型分数是怎么从输入算出来的？**

先从一个线性层开始。你之前算过学生综合成绩：

```python
scores @ weights
```

线性层在这个基础上加了一个**偏置（bias）**：

```text
z = x @ w + b
```

- `x`：输入，例如三门课的成绩。
- `w`：权重，决定每个输入如何影响输出。
- `b`：偏置，提供一个不依赖输入的基础值。
- `z`：计算结果，此时不一定是概率。

例如：

```python
import numpy as np

x = np.array([80, 90, 70])
w = np.array([0.5, 0.3, 0.2])
b = 5.0

z = x @ w + b
print(z)  # 86.0
```

原来的加权成绩是 `81`，加上偏置 `5`，得到 `86`。这里的成绩例子用于说明计算结构。

在神经网络中，**权重和偏置都是可训练的参数**。我们先手动给它们赋值，后面再学习如何根据 loss 调整它们。

一个线性层也可以同时计算多个输出。假设输入有三个特征，要输出两个分数：

```text
x 的形状：   (3,)
W 的形状：   (3, 2)
b 的形状：   (2,)

z = x @ W + b
z 的形状：   (2,)
```

`W` 的每一列对应一个输出的权重：

```python
x = np.array([1.0, 2.0, 3.0])

W = np.array([
    [1.0,  0.0],
    [0.0,  1.0],
    [1.0, -1.0],
])

b = np.array([0.5, 1.0])

z = x @ W + b
```

先不用运行，试着手算：**`z` 的两个值分别是多少？** 提示：分别用 `x` 与 `W` 的两列做点积，再加上对应的偏置。
