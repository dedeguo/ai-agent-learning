# 04 梯度、学习率与训练循环 

我们进入第一周第四课：**梯度下降与参数更新**。

你已经会计算：

```text
输入 → 模型 → 预测 → loss
```

接下来要解决：**知道 loss 后，怎样调整参数，让预测更准确？**

先用只有一个参数的模型：

```text
预测值 ŷ = w × x
```

假设输入 `x = 2`，真实答案 `y = 6`。我们希望模型学到合适的 `w`，而不是直接把答案告诉它。

初始设 `w = 1`：

```text
预测值 = 1 × 2 = 2
loss = (预测值 - 真实值)²
     = (2 - 6)² = 16
```

直觉上，增大 `w` 会让预测更接近 `6`。但程序需要一个计算规则来判断方向，这就是**梯度**。

对于一个参数，梯度就是损失对这个参数的导数。它表示：**参数发生很小变化时，损失变化的方向和快慢。**

本例中：

```text
loss = (wx - y)²

梯度 = d(loss)/dw
     = 2(wx - y) × x
```

暂时不用急着推导，先代入数值：

```text
w = 1，x = 2，y = 6

梯度 = 2 × (1×2 - 6) × 2
     = -16
```

**梯度为负，意味着在当前位置稍微增大 `w`，loss 会下降。**

参数更新规则是：

```text
新参数 = 旧参数 - 学习率 × 梯度
```

学习率控制每次调整的幅度。设学习率为 `0.1`：

```text
新 w = 1 - 0.1 × (-16)
     = 2.6
```

更新后：

```text
预测值 = 2.6 × 2 = 5.2
loss = (5.2 - 6)² = 0.64
```

损失从 `16` 降到 `0.64`，这就是一次梯度下降更新。

先做一个方向判断：如果当前 **`w = 4`**，输入仍是 `2`，真实答案仍是 `6`，那么预测是 `8`。你觉得此时梯度是正还是负？更新应该让 `w` 增大还是减小？

````python

x=2.0
y=6.0
w=1.0
learning_rate=0.2

for step in range(20):
    prediction = w * x
    loss =(prediction-y)**2
    gradient = 2* (prediction-y)*x
    print(
        f"第 {step} 步：w={w:.6f},"
        f"预测={prediction:.6f}, loss={loss:.6f}"
    )
    w=w-learning_rate*gradient
````

````python
x=2.0
y=6.0

w=1.0
b=1.0
learning_rate=0.1

for step in range(20):
    prediction = w * x + b
    loss =(prediction-y)**2
    gradient_w = 2* (prediction-y)*x
    gradient_b = 2* (prediction-y)
    print(
        f"第 {step} 步：w={w:.6f}, b={b:.6f}"
        f"预测={prediction:.6f}, loss={loss:.6f}"
    )
    w=w-learning_rate*gradient_w
    b=b-learning_rate*gradient_b
````

````python
x, y = 2.0, 6.0
w, b = 1.0, 1.0
learning_rate = 0.05

for step in range(10):
    prediction = w * x + b
    error = prediction - y
    loss = error ** 2

    gradient_w = 2 * error * x
    gradient_b = 2 * error

    print(f"{step}: w={w:.4f}, b={b:.4f}, loss={loss:.6f}")

    w -= learning_rate * gradient_w
    b -= learning_rate * gradient_b
````

````python
import numpy as np

x = np.array([1.0, 2.0, 3.0])
y = np.array([3.0, 5.0, 7.0])

w, b = 0.0, 0.0
learning_rate = 0.05

for step in range(1000):
    # 前向计算：同一套 w、b 预测三个样本
    prediction = w * x + b
    error = prediction - y

    # 平均平方误差
    loss = np.mean(error ** 2)

    # 平均损失对两个参数的梯度
    gradient_w = np.mean(2 * error * x)
    gradient_b = np.mean(2 * error)

    if step % 100 == 0:
        print(f"{step}: w={w:.4f}, b={b:.4f}, loss={loss:.6f}")

    # 两个梯度都算好后，再更新参数
    w -= learning_rate * gradient_w
    b -= learning_rate * gradient_b

print("最终参数：", w, b)
print("最终预测：", w * x + b)
````
