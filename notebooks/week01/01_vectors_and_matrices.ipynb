```python
# 标量：一个数字
score = 3.0

# 向量：一列有顺序的数字
student = [80, 90, 70]

# 矩阵：多行、多列的数字
students = [
    [80, 90, 70],
    [60, 75, 85],
]
```

```python
x=np.array([2,3,4])
w=np.array([1,0,2])
print("点积：",x@w)

scores = np.array(
    [
        [80,90,70],
        [60,75,85],
    ]
)
print("原始形状：",scores.shape)
print("转置后的形状：",scores.T.shape)
print("转置后的成绩：",scores.T)
```

```python
w=np.array([0.5,0.3,0.2])
overall_score=scores @ w
print("学生综合成绩：",overall_score)
```

# 余弦相似度

```python
a=np.array([0.5,0.3])
b=np.array([0,1])
dot_ab=a @ b
print("ab 点积:",dot_ab)
cos_ab = dot_ab / (np.linalg.norm(a) * np.linalg.norm(b))
print("ab 余弦相似度:",cos_ab)
```
